Methods and its types:

Methods are of 3 types: 1.Instance method 2.Class method 3. Static method

```python
#Instance/object method:
class Person:
  def __init__(self,name,age):
    self.name = name
    self.age = age

  def findAge(self):
    return self.age

per = Person("Swastika",99)
print(per.findAge())

```

```python
#class method

class Person:
  country = "India"

  @classmethod
  def greet(cls):
    print("Hello from the", cls.country)

  def __init__(self,name,age):
    self.name = name
    self.age = age

  def findAge(self):
    return self.age

per = Person("Swastika",99)
print(per.findAge())

Person.greet()
```

```python
#class method

class Person:
  country = "India"

  @classmethod
  def greet(cls):
    print("Hello from the", cls.country)

  def __init__(self,name,age):
    self.name = name
    self.age = age

  def findAge(self):
    return self.age

per = Person("Swastika",99)
print(per.findAge())

Person.greet()
per.greet()
```
