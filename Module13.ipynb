Initializer and Constructor

```python
class Person:
  name = "Swastika"
  age = 99

per1 = Person()
per2 = Person()
per3 = Person()

print(per1.name)
print(per2.name)
print(per3.name)
```

```python
class Person:
  #initializer and constructor for objects
  def __init__(self,name,age):
    self.name = name
    self.age = age

per1 = Person("Swastika",99)
per2 = Person("Shivani",97)

print(per1.name)
print(per2.name)
```
