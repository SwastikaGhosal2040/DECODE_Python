Initializer and Constructor

```python
class Person:
  name = "Swastika"
  age = 99

per1 = Person()
per2 = Person()
per3 = Person()

print(per1.name)
print(per2.name)
print(per3.name)
```

```python
class Person:
  #initializer and constructor for objects
  def __init__(self,name,age):
    self.name = name
    self.age = age

per1 = Person("Swastika",99)
per2 = Person("Shivani",97)

print(per1.name)
print(per2.name)
```

```python
class Person:
  def __init__(self,name,age): #ignored
    self.name = name
    self.age = age

  def __init__(self,name): #actual init method
    self.name = name

  per = Person("Swastika Ghosal")
  print(per.name)
```

```python
class Person:
  def __init__(self,name,age): #ignored
    self.name = name
    self.age = age

  def __init__(self,name): #actual init method
    self.name = name

  per = Person("Swastika Ghosal",99)
  print(per.name)
```

```python
class Person:
  def __init__(self,name,age=99,hobby="Cricket"):
    #as value of age and hobby are provided, they are optional parameters
    self.name = name
    self.age = age
    self.hobby = hobby

per1 = Person("Swastika")
per2 = Person("Swastika",99)
per3 = Person("Swastika",43,"Football")

print(per1.hobby)
print(per3.hobby)
```

```python
class Person:
  country = "India" #attribute of class should always be initialized(same for all objects)
  def __init__(self,name,age):
    self.name = name #instance attribute keep changing for all objects
    self.age = age

per1 = Person("Swastika", 99)
per2 = Person("Shweta", 73)

print(per1.name,per1.country)
print(per2.name,per2.country)

print(Person.country) #it can be directly called
```
